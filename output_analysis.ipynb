# Output analysis

```python
import os

import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

```python
items = []
for file in os.listdir("output"):
    items.append(os.path.basename(file))
items.sort()
```

```python
for item in items:
    fn = f'output/{item}/out_partial_dependence_age_{item}.png'
    img = mpimg.imread(fn)
    plt.imshow(img)
    plt.axis('off') # Hide coordinate axes
    plt.show()
```

```python
for item in items:
    fn = f'output/{item}/out_partial_dependence_sex_{item}.png'
    img = mpimg.imread(fn)
    plt.imshow(img)
    plt.axis('off') # Hide coordinate axes
    plt.show()
```

```python
for item in items:
    fn = f'output/{item}/out_shap_boxplots_{item}.png'
    img = mpimg.imread(fn)
    plt.imshow(img)
    plt.axis('off') # Hide coordinate axes
    plt.show()
```

```python
from matplotlib.backends.backend_pdf import PdfPages

with PdfPages('summary_fourier_regression.pdf') as pdf:

    for item in items:

        labels =['age', 'sex', 'shap']

        image_paths = [
            f'output/{item}/out_partial_dependence_age_{item}.png',
            f'output/{item}/out_partial_dependence_sex_{item}.png',
            f'output/{item}/out_shap_boxplots_{item}.png'
            ]

        fig, axes = plt.subplots(1, 3, figsize=(8, 5))

        for i, img_path in enumerate(image_paths):
            img = mpimg.imread(img_path)
            axes[i].imshow(img)
            axes[i].set_title(labels[i], fontsize=8)
            axes[i].axis('off')
        
        fig.suptitle(item, fontsize=10, y=0.7)
        
        pdf.savefig(fig, dpi=300)
        plt.close(fig)
```

```python
for item in items:
    fn = f'output/{item}/out_performance_{item}.png'
    img = mpimg.imread(fn)
    plt.imshow(img)
    plt.axis('off') # Hide coordinate axes
    plt.title(item, fontsize=10)
    plt.show()
```
